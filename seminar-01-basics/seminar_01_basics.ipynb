## 1. Variables and numbers

```python
a = 3
print(a)
```

```python
a = 4
print(a)
```

```python
a = a + 5
print(a)
```

```python
print('a =', a)
```

```python
print('a + 2 =', a + 2)
print('a - 2 =', a - 2)
print('a * 2 =', a * 2)
print('a / 2 =', a / 2)
print('a // 2 =', a // 2)
print('a % 2 =', a % 2)
print('a ** 2 =', a ** 2)
```

```python
print(4 / 2)
print(4 // 2)
```

```python
print(17 // 5)
print(17 % 5)
print(2 ** 10)
x = 10 
x = x * 2 
print(x)
```
