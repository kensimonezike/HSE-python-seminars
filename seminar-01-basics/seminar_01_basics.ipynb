## What I took from this seminar

- `=` stores a value in a variable. `==` asks whether two things are equal.
- `/` always gives a decimal, even for `4 / 2`. `//` throws away the decimals and `%` gives the remainder. `n % 2` tells me if a number is even.
- `input()` always returns text, even when I type a number. I wrap it in `int()` or `float()` before doing maths, otherwise Python raises a `TypeError`.
- `'1' + 2` fails because one is text and the other is a number. The error message names both types, so reading it tells me the fix.
- The last digit of a whole number is `n % 10`. The course answer key used `n % 100 // 10`, which gives the tens digit, so I checked it by running it.
- f-strings like `f"{name} is {age} years old"` are easier to read than joining pieces with `+`, and they don't need `str()`.
- String methods such as `strip()`, `upper()`, `replace()` and `split()` return a new string. The original stays the same unless I store the result.
